# Primeiro DataSet
## Fonte UCI : Appliances Energy Prediction
### Consumo de eletrodomésticos em uma residência, associado a temperatura, umidade e condições ambientais.

```python
energydata = pd.read_csv('energydata_complete.csv')
Rn_energydata = energydata.rename(columns={
    "Appliances" : "Consumo_Eletrodomesticos",
    "T_out": "temp_ext",
    "RH_out": "umid_ext",
    "Windspeed":"vento",
})
```

```python
Rn_energydata.head(5)
Rn_energydata.shape[0]
Rn_energydata.info()
Rn_energydata.describe()
```

```python
consumo_max = Rn_energydata["Consumo_Eletrodomesticos"].max()
print(f"Maior consumo de eletrodomésticos: {consumo_max}")
```

```python
limiar_consumo = 0.7 * consumo_max
df_consumo_alto = Rn_energydata[Rn_energydata["Consumo_Eletrodomesticos"] > limiar_consumo]

print(f"Limiar (70% do máximo): {limiar_consumo}")
df_consumo_alto.head()
```

```python
n_selecionados = df_consumo_alto.shape[0]
n_total = Rn_energydata.shape[0]
percentual = (n_selecionados / n_total) * 100

print(f"Registros acima do limiar: {n_selecionados}")
print(f"Total de registros na amostra: {n_total}")
print(f"Percentual da amostra: {percentual:.2f}%")
```

```python
temp_media_T1 = Rn_energydata["T1"].mean()

df_consumo_e_temp = Rn_energydata[
    (Rn_energydata["Consumo_Eletrodomesticos"] > limiar_consumo) &
    (Rn_energydata["T1"] > temp_media_T1)
]

n_duplo_criterio = df_consumo_e_temp.shape[0]
percentual_duplo = (n_duplo_criterio / n_total) * 100

print(f"Temperatura média (T1): {temp_media_T1:.2f}")
print(f"Registros com consumo alto E temperatura acima da média: {n_duplo_criterio}")
print(f"Percentual da amostra: {percentual_duplo:.2f}%")

df_consumo_e_temp.head()
```

```python
comparacao = pd.DataFrame({
    "Critério": ["Consumo > 70% do máximo", "Consumo > 70% E T1 > média"],
    "Nº de registros": [n_selecionados, n_duplo_criterio],
    "% da amostra": [round(percentual, 2), round(percentual_duplo, 2)]
})

comparacao
```

- Primeiro Critério (Somente Consumo Alto): Ao selecionar apenas registros onde o 'Consumo_Eletrodomesticos' era superior a 70% do máximo (756.0), foram identificados 31 registros, o que representa 0.16% da amostra total.

- Segundo Critério (Consumo Alto E Temperatura Acima da Média): Quando adicionamos a condição de que a temperatura (T1) também deveria estar acima da média (21.69), o número de registros diminuiu para 14, correspondendo a 0.07% da amostra total.

##### Efeito da Inclusão da Temperatura: A inclusão da temperatura como segundo critério teve o efeito de reduzir significativamente o número de registros selecionados, de 31 para 14. Isso indica que, embora houvesse períodos de alto consumo de eletrodomésticos, nem todos eles ocorreram quando a temperatura interna (T1) estava acima da média. Ao adicionar essa condição, estamos filtrando para um cenário mais específico que seriam momentos de alto consumo que são acompanhados por temperaturas internas mais elevadas. Isso pode ser útil para identificar padrões de uso de energia em condições climáticas ou ambientais específicas, por exemplo, o uso de aquecedores ou ar-condicionado em conjunto com outros eletrodomésticos.

# Escolhendo o Lag_Potencia.Reactiva_kVarh e definindo o parametro para saber os valores baixos dele

```python
media_lag_fator_potencia = dados_maximum_load["Lag_Fatores.de.Potencia"].mean()
print(f"Media dos valores de Lag_Fatores.de.Potencia: {media_lag_fator_potencia:.3f}")
print(f"Portanto o valor do parametro vai ser {media_lag_fator_potencia:.3f}")
```

### Registros com Consumo Elevado e Fator de Potência Baixo

Agora, vamos criar um novo DataFrame que combina as duas condições: consumo elevado (que já está em `dados_maximum_load`) e um fator de potência (Lag_Fatores.de.Potencia) abaixo da media

```python
dados_potencia_baixa_consumo_alto = dados_maximum_load["Lag_Fatores.de.Potencia"] < media_lag_fator_potencia
dados_potencia_baixa_consumo_alto = dados_maximum_load[dados_potencia_baixa_consumo_alto]
dados_potencia_baixa_consumo_alto.head()
```

### Por que esses registros merecem mais atenção?

Este ultimo DataFrame (`dados_potencia_baixa_consumo_alto`), que apresenta **alto consumo de energia ('Maximum_Load')** simultaneamente com **baixos fatores de potência (`Lag_Fatores.de.Potencia` < 90.613)**, é crítico pelos seguintes motivos:

1. O baixo fator de potência significa que os equipamentos estão consumindo mais corrente do que o necessário para fazer o trabalho , o que se gera um maior desperdício de energia e custos operacionais mais altos.

2.  A alta corrente e o baixo fator de potência pode gerar uma sobrecarregar na infraestrutura elétrica (transformadores, disjuntores, cabos), reduzindo sua vida útil e aumentando o risco de erros e parando a produção.

3. Esse tipo de problemas , assim como pode trazer muitos problemas para a empresa também ajuda a verificar e aprender dos erros que antes eram desconhecidos , fazendo com que seja mais facil arrumar as falhas no futuro , uma vez que já sabemos como soluciona isso.

# Terceiro DataSet
## Fonte UCI: Power Consumption of Tetouan City
### Consumo elétrico de três zonas de distribuição da cidade de Tétouan, associado a variáveis meteorológicas.

```python
tetuan = pd.read_csv('Tetuan City power consumption.csv')

Rn_tetuan = tetuan.rename(columns={
    "Zone 1 Power Consumption": "Consumo_Zona_1",
    "Zone 2  Power Consumption": "Consumo_Zona_2",
    "Zone 3  Power Consumption": "Consumo_Zona_3",
})

Rn_tetuan.head(5)
```

```python
max_zona1 = Rn_tetuan["Consumo_Zona_1"].max()
max_zona2 = Rn_tetuan["Consumo_Zona_2"].max()
max_zona3 = Rn_tetuan["Consumo_Zona_3"].max()

print(f"Consumo máximo - Zona 1: {max_zona1:.2f}")
print(f"Consumo máximo - Zona 2: {max_zona2:.2f}")
print(f"Consumo máximo - Zona 3: {max_zona3:.2f}")
```

```python
maximos = {
    "Zona 1": max_zona1,
    "Zona 2": max_zona2,
    "Zona 3": max_zona3,
}

zona_pico = max(maximos, key=maximos.get)
consumo_max = maximos[zona_pico]

print(f"A zona com maior pico de consumo é: {zona_pico}")
print(f"Valor do pico: {consumo_max:.2f}")
```

```python
# Mapeia o nome da zona vencedora para a coluna correspondente
coluna_pico = {
    "Zona 1": "Consumo_Zona_1",
    "Zona 2": "Consumo_Zona_2",
    "Zona 3": "Consumo_Zona_3",
}[zona_pico]

limiar_consumo = 0.7 * consumo_max

df_consumo_alto = Rn_tetuan[Rn_tetuan[coluna_pico] > limiar_consumo]

print(f"Limiar (70% do máximo da {zona_pico}): {limiar_consumo:.2f}")
df_consumo_alto.head()
```

```python
n_selecionados = df_consumo_alto.shape[0]
n_total = Rn_tetuan.shape[0]
percentual = (n_selecionados / n_total) * 100

print(f"Registros acima do limiar: {n_selecionados}")
print(f"Total de registros na amostra: {n_total}")
print(f"Percentual da amostra: {percentual:.2f}%")
```

```python
temp_media = Rn_tetuan["Temperature"].mean()

df_consumo_e_temp = Rn_tetuan[
    (Rn_tetuan[coluna_pico] > limiar_consumo) &
    (Rn_tetuan["Temperature"] > temp_media)
]

n_duplo_criterio = df_consumo_e_temp.shape[0]
percentual_duplo = (n_duplo_criterio / n_total) * 100

print(f"Temperatura média: {temp_media:.2f}")
print(f"Registros com consumo alto E temperatura acima da média: {n_duplo_criterio}")
print(f"Percentual da amostra: {percentual_duplo:.2f}%")

df_consumo_e_temp.head()
```

```python
comparacao = pd.DataFrame({
    "Critério": [f"Consumo > 70% do máximo ({zona_pico})", "Consumo > 70% E Temperatura > média"],
    "Nº de registros": [n_selecionados, n_duplo_criterio],
    "% da amostra": [round(percentual, 2), round(percentual_duplo, 2)]
})

comparacao
```

Com base nos resultados:

- Primeiro Critério (Somente Consumo Alto): Ao selecionar registros onde o consumo na Zona 1 (a zona de pico) estava acima de 70% do seu máximo (36543.08), foram identificados 14718 registros, o que representa 28.08% da amostra total.

- Segundo Critério (Consumo Alto E Temperatura Acima da Média): Ao adicionar a condição de que a temperatura também deveria estar acima da média (18.81°C), o número de registros diminuiu para 9917, correspondendo a 18.92% da amostra total.

- Efeito da Inclusão da Condição Ambiental (Temperatura): A inclusão da temperatura como segundo critério teve o efeito de reduzir significativamente o número de registros selecionados. Houve uma diminuição de 14718 para 9917 registros. Isso indica que, embora houvesse muitos períodos de alto consumo de energia na Zona 1, uma parte grande desses eventos não aconteceu ao mesmo tempo com temperaturas acima da média. Ao adicionar essa condição, estamos filtrando para um cenário mais específico, mostrando que o alto consumo nem sempre tem uma conexão direta com temperaturas elevadas neste dataset, ou que há outros fatores predominantes em alguns picos de consumo.

## Fonte Kaggle : Wind & Solar Energy Production Dataset
### Produção de energia eólica e solar, permitindo comparar o comportamento das duas fontes renováveis.

```python
dados = pd.read_csv("Energy Production Dataset.csv", sep=",")
```

```python
dados = dados.rename(columns={ "Production": "producao",

                              "Source": "Fonte",

                              "Date": "Data",

                              "Season": "Temporal" })
```

```python
dados['Fonte'] = dados['Fonte'].replace({'Solar': 'Geracao_Solar', 'Wind': 'Geracao_Eolica'})



dados= dados.sample(frac=0.7)



dados = dados [["Data" , "Fonte" , "Temporal", "producao"]]



print(dados)
```

```python
dados_eolica = dados.query('Fonte=="Geracao_Eolica"').sort_values('producao')

dados_Solar = dados.query('Fonte=="Geracao_Solar"').sort_values('producao')
```

```python
maximo_solar =  dados_Solar["producao"].max()

minimo_solar = dados_Solar["producao"].min()



maximo_eolica =  dados_eolica["producao"].max()

minimo_eolica = dados_eolica["producao"].min()



corte_solar = maximo_solar * 0.7

corte_eolica = maximo_eolica * 0.7
```

```python
alta_solar = dados_Solar.query('producao > @corte_solar').sort_values('producao')

alta_eolica = dados_eolica.query('producao > @corte_eolica').sort_values('producao')



qtd_alta_solar = len(alta_solar)

qtd_alta_eolica = len(alta_eolica)



pct_alta_solar = (qtd_alta_solar / len(dados_Solar)) * 100

pct_alta_eolica = (qtd_alta_eolica / len(dados_eolica)) * 100
```

```python
print(f"Solar  -> Registros > 70%: {qtd_alta_solar} ({pct_alta_solar:.2f}%)")

print(f"Eólica -> Registros > 70%: {qtd_alta_eolica} ({pct_alta_eolica:.2f}%)\n")



if pct_alta_solar > pct_alta_eolica:

    print("A fonte Solar aparece com maior frequência acima de 70% do seu máximo.")

elif pct_alta_eolica > pct_alta_solar:

    print("A fonte Eólica aparece com maior frequência acima de 70% do seu máximo.")

else:

    print("Ambas as fontes apresentaram a mesma frequência relativa acima de 70%.")
```

- Diferenças de escala: Fontes solar e eólica possuem capacidades máximas distintas; um limite fixo representaria proporções totalmente diferentes do potencial de cada uma.
- Distorção de desempenho: Favoreceria artificialmente a fonte com maior volume total de geração, mascarando a frequência real com que cada uma atinge seu próprio topo.
- Dinâmicas operacionais opostas: Ignora as características físicas e as curvas de geração específicas de cada tecnologia (como a intermitência solar em comparação aos regimes de vento).
##### No seu código, esse problema é evitado ao utilizar limites relativos (calculados como uma porcentagem do máximo de cada fonte individualmente), garantindo uma comparação justa.

# Sexto DataSet
## Fonte UCI: Household Eletric Power Consumption
### Medições de consumo elétrico residencial, tensão, corrente, potência ativa, potência reativa e submedições.

```python
# Carregar o arquivo
df = pd.read_csv("household_power_consumption.csv")

# Renomear as colunas
df = df.rename(columns={
    "Global_active_power": "potencia_ativa",
    "Global_reactive_power": "potencia_reativa",
    "Voltage": "tensao",
    "Global_intensity": "corrente",
    "Sub_metering_1": "medição_1",
    "Sub_metering_2": "medicao_2",
    "Sub_metering_3": "medicao_3"
})
```

```python
print(df.head())

print(df.columns)
```

### Determinar o valor máximo de potência ativa

```python
potencia_maxima = df["potencia_ativa"].max()

print("Potência ativa máxima:", potencia_maxima)
```

## Calcular 75% do máximo e criar o primeiro DataFrame

```python
limite_75 = potencia_maxima * 0.75

print("Limite de 75%:", limite_75)

df_75 = df[df["potencia_ativa"] > limite_75]

df_75.head()
```

## Calcular quantidade e percentual de registros selecionados

```python
quantidade_75 = len(df_75)

print("Quantidade de registros:", quantidade_75)

percentual_75 = (quantidade_75 / len(df)) * 100

print("Percentual de registros:", percentual_75, "%")

print(f"Percentual de registros: {percentual_75:.4f}%")
```

## Calcular a corrente média da amostra

```python
corrente_media = df["corrente"].mean()

print("Corrente média:", corrente_media)

print(f"Corrente média: {corrente_media:.2f} A")
```

## Criar o segundo DataFrame com as duas condições

```python
df_75_corrente = df[
    (df["potencia_ativa"] > limite_75) &
    (df["corrente"] > corrente_media)
]

print(df_75_corrente.head())

quantidade_2 = len(df_75_corrente)

print("Quantidade de registros:", quantidade_2)

percentual_2 = (quantidade_2 / len(df)) * 100

print(f"Percentual: {percentual_2:.4f}%")
```

## Comparar os dois conjuntos

```python
print("Registros somente pela potência:", len(df_75))
print("Registros por potência + corrente:", len(df_75_corrente))

print("Registros eliminados pela segunda condição:",
      len(df_75) - len(df_75_corrente))

proporcao = (len(df_75_corrente) / len(df_75)) * 100

print(f"{proporcao:.2f}% dos registros do primeiro conjunto também atendem à condição de corrente.")
```

#### Comparação dos conjuntos
- Primeiro conjunto: foram selecionados 58 registros com potência ativa acima de 75% do valor máximo.
- Segundo conjunto: foram aplicadas duas condições simultaneamente: potência ativa acima de 75% do máximo e corrente acima da média de 4,63 A.
- Resultado: o segundo conjunto também possui 58 registros.
- Efeito da nova condição: nenhum registro foi eliminado, pois todos os registros que apresentaram potência acima de 75% também possuíam corrente acima da média.
##### Conclusão: a inclusão da corrente como segunda condição não alterou a seleção dos registros nesta amostra, mantendo os mesmos 58 casos.
